```python
#7. Data Cleaning
#Handle Missing Values
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())
```

```python
df.isnull().sum()
```

```python
#Clean Missing Values and Remove Duplicates
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df.groupby('Season')['Soil_Moisture_pct'].transform('median'))
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df.groupby('Season')['Yield_Tonnes_Ha'].transform('median'))
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df.groupby('Season')['Rainfall_mm'].transform('median'))

df.drop_duplicates(inplace=True)

print("Missing values after cleaning:")
print(df[['Soil_Moisture_pct', 'Yield_Tonnes_Ha', 'Rainfall_mm']].isnull().sum())
```

```python
#Statistical Significance
from scipy.stats import f_oneway
kharif = df[df["Season"] == "Kharif"]["Yield_Tonnes_Ha"].dropna()
rabi = df[df["Season"] == "Rabi"]["Yield_Tonnes_Ha"].dropna()
zaid = df[df["Season"] == "Zaid"]["Yield_Tonnes_Ha"].dropna()

f_stat, p_value = f_oneway(kharif, rabi, zaid)

print("F-statistic:", f_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("The difference in average yield across seasons is statistically significant.")
else:
    print("The difference in average yield across seasons is not statistically significant.")
```

```python
Q1 = df["Yield_Tonnes_Ha"].quantile(0.25)
Q3 = df["Yield_Tonnes_Ha"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Yield_Tonnes_Ha"] < lower_bound) |
    (df["Yield_Tonnes_Ha"] > upper_bound)
]

print("Number of yield outliers:", len(outliers))
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```
